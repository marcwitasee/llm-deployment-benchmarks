```sql
CREATE TABLE IF NOT EXISTS support.bugzilla.urgency_set AS
SELECT * FROM (
  SELECT *, ROW_NUMBER() OVER (PARTITION BY is_urgent ORDER BY rand()) AS rn
  FROM support.bugzilla.silver_bugs
  WHERE id NOT IN (SELECT id FROM support.bugzilla.eval_set)
  AND id NOT IN (SELECT id FROM support.bugzilla.eval_set_2)
)
WHERE rn <= 100
```
