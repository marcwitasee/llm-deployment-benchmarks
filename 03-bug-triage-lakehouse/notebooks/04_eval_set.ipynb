```sql
CREATE TABLE IF NOT EXISTS support.bugzilla.eval_set AS
SELECT * FROM (
  SELECT *, ROW_NUMBER() OVER (PARTITION BY team ORDER BY rand()) AS row_number
  FROM support.bugzilla.silver_bugs
)
WHERE row_number < 21
```

```sql
SELECT team, COUNT(*) FROM support.bugzilla.eval_set GROUP BY team
```
